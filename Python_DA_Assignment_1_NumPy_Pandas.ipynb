## Part 1: NumPy Array Operations

### 1. Create a 1D NumPy Array
Create the Week 1 temperature array.

```python
import numpy as np
import pandas as pd

# Temperatures for Week 1 (in Celsius)
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 temperatures:", temperatures_w1)
```

### 2. Inspection and Properties
Display the shape, data type, and number of elements.

```python
print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

### 3. Array Operations
Convert Celsius to Fahrenheit and calculate the maximum, minimum, and mean temperatures.

```python
# Celsius to Fahrenheit: (Celsius * 9/5) + 32
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:", temperatures_f)
print("Maximum temperature:", np.max(temperatures_w1))
print("Minimum temperature:", np.min(temperatures_w1))
print("Mean temperature:", np.mean(temperatures_w1))
```

### 4. Array Slicing and Indexing
- First three days
- Weekend (last two days)
- Middle three days

```python
print("First three days:", temperatures_w1[:3])
print("Weekend temperatures:", temperatures_w1[-2:])
print("Middle three days:", temperatures_w1[2:5])
```

### 5. Create a 2D Array
Each row represents one week.

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Temperatures for both weeks:")
print(temperatures)
```

### 6. Inspect and Slice the 2D Array
Display its properties, each week's temperatures, and weekend temperatures for both weeks.

```python
print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total number of elements:", temperatures.size)

print("Week 1:", temperatures[0])
print("Week 2:", temperatures[1])
print("Week 1 weekend:", temperatures[0, -2:])
print("Week 2 weekend:", temperatures[1, -2:])
```

## Part 2: Pandas Series

### 1. Creating a Pandas Series
Create a Series named `marks` with custom index labels.

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"]
)

print(marks)
```

### 2. Indexing and Slicing
Access marks using integer position, `loc`, `iloc`, and a Boolean condition.

```python
# Integer position: first rank student
print("First rank mark:", marks.iloc[0])

# loc: marks of the top three ranks using labels
print("\nTop three ranks:")
print(marks.loc[["Rank1", "Rank2", "Rank3"]])

# iloc: third rank student (position 2)
print("\nThird rank mark:", marks.iloc[2])

# Boolean mask: marks greater than 90
print("\nRanks with marks greater than 90:")
print(marks[marks > 90])
```

### 3. Manipulating Series
Change Rank1 to 100, remove Rank5, and calculate CGPA by dividing marks by 10.

```python
# Modify the first-rank student's mark
marks.loc["Rank1"] = 100

# Remove the last-rank entry
marks = marks.drop("Rank5")

# Calculate CGPA
cgpa = marks / 10

print("Updated marks:")
print(marks)

print("\nCGPA:")
print(cgpa)
```

## Part 3: Pandas DataFrame

### 1. Creating a Pandas DataFrame
Create the transactions DataFrame using the values in the assignment.

```python
transactions = pd.DataFrame({
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory": [
        "Electronics", "Clothing", "Electronics", "Furniture", "Clothing",
        "Electronics", "Furniture", "Clothing", "Furniture", "Electronics"
    ],
    "Region": [
        "North", "South", "North", "East", "West",
        "North", "East", "West", "South", "North"
    ],
    "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

### 2. Data Exploration
Explore the DataFrame and perform the requested selections, filtering, counts, unique-value check, and grouping.

```python
# Display first five rows
print("First five rows:")
print(transactions.head())

# Display last five rows
print("\nLast five rows:")
print(transactions.tail())

# Shape, column names, and data types
print("\nShape:", transactions.shape)
print("\nColumn names:", transactions.columns.tolist())
print("\nData types:")
print(transactions.dtypes)

# Display only ProductCategory and Amount
print("\nProductCategory and Amount:")
print(transactions[["ProductCategory", "Amount"]])

# Retrieve the last three columns
print("\nLast three columns:")
print(transactions.iloc[:, -3:])

# Filter: Region is North and Amount is greater than 200
print("\nNorth region with Amount greater than 200:")
print(transactions[(transactions["Region"] == "North") & (transactions["Amount"] > 200)])

# Count each product category
print("\nProduct category counts:")
print(transactions["ProductCategory"].value_counts())

# Find unique regions
print("\nUnique regions:")
print(transactions["Region"].unique())

# Mean amount for each region
print("\nMean amount by region:")
print(transactions.groupby("Region")["Amount"].mean())
```

### 3. Manipulating the DataFrame
Update TransactionID 102, add a 10% Discount column, remove TransactionID 109, and then delete the Discount column.

```python
# Change Amount for TransactionID 102 to 165
transactions.loc[transactions["TransactionID"] == 102, "Amount"] = 165

# Add Discount column (10% of Amount)
transactions["Discount"] = transactions["Amount"] * 0.10

print("After updating Amount and adding Discount:")
print(transactions)

# Remove the row with TransactionID 109
transactions = transactions[transactions["TransactionID"] != 109]

print("\nAfter removing TransactionID 109:")
print(transactions)

# Delete the Discount column
transactions = transactions.drop(columns=["Discount"])

print("\nFinal DataFrame after deleting Discount:")
print(transactions)
```
